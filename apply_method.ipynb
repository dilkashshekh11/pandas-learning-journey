```python
# apply method part - 1
import pandas as pd
billing = pd.Series([450.75, 1200.50, 980.00,1550.25, 760.40, 2300.00, 890.10, 1475.60, 650.00,3100.75], name="billing")
billing
```

```python
def give_discount(bill):
    if bill >= 1000:
        return bill - (bill * 10/100)
    return bill
```

```python
billing.apply(give_discount)
# round(billing.apply(give_discount),2)
```

```python
# Apply method part - 2

students = ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Sana", "Vikram", "Zoya"]
mark_series = pd.Series ([92, 78, 35, 66, 88, 49, 57, 95], index = students, name="marks")
```

```python
def grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "F"
```

```python
grades_reuslt = mark_series.apply(grade)
grades_reuslt
```

```python
grades_reuslt.value_counts()
```
