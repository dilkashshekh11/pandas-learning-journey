```python
import numpy as np
import pandas as pd

department = ["engineering", "marketing", "sales", "hr", "finance", "support",
"design", "operations", "legal", "research", "admin", "security"]

salary_series = pd.Series([52.0, 47.5, 61.0, np.nan, 58.5, 43.0,
49.5, 54.0, np.nan, 66.0, 41.5, np.nan], index = department, name="salary")
salary_series 
```

```python
salary_series.count()
```

```python
salary_series.size # also inclue non values
```

```python
# salary_series.isna() # this code only give true or false values
salary_series.isna().sum() # count non value (sum count true values)
```

```python
salary_series.sum()
```

```python
salary_series.mean()  #( sum / divided by all non null values)
```

```python
salary_series.median()
```

```python
salary_series.min()
```

```python
salary_series.max()
```

```python
# skipna = False stops pandas from ignoring NaN
# salary_series.mean()
salary_series.mean(skipna = False) # also divide by all number of element ( even value are empty or non)
```

```python
# filling NaN with 0 pulls the average down, which is why missing is kept as NaN
salary_series.fillna(0).mean()
```

```python
round(salary_series.mean(),2)
```
