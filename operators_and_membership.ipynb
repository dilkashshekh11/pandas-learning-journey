```python
import numpy as np
import pandas as pd

salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series (salary, index = department, name = "Salary") 
salary_series
```

```python
print('(salary_series == 0) = ',salary_series == 0)
print('(salary_series != 0) = ',salary_series != 0)
print('(salary_series > 50000) = ',salary_series > 50000)
print('(salary_series >= 50000) = ',salary_series >= 50000)
print('(salary_series < 50000) = ',salary_series < 50000)
print('(salary_series <= 48000) = ',salary_series <= 48000)
```
