```python
import pandas as pd
billing  = pd.Series([450.75, 1200.50, 980.00, 1550.25, 760.40, 2300.00, 890.10, 1475.60, 650.00, 3100.75],name="billing")
```

```python
# Apply 10% discount to bills > 1000
final_price = billing.where(billing <= 1000,(billing * 0.90))
# final_price = billing.where(billing <= 1000,(billing * 0.90).round(2))
final_price
```

```python
# Apply 15% discount to bills > 1000
final_price2 = billing.where(billing <= 1000, (billing *(0.85)))
final_price2
```
