```python
import pandas as pd
salaries = [45000, 52000, 0, 61000, 58000, 72000, 49000, 65000]
departments = ["developer", "marketing", "hr", "sales", "finance", "support", "design", "operations"]
salary_series = pd.Series (salaries, index = departments, name="Salary")
print(salary_series)
salary_series[:5:2]
```

```python
# Indexing
print(f'salary_series["developer"] = {salary_series["developer"]}')
print(f'salary_series["sales"] = {salary_series["sales"]}')
print(f'salary_series[["sales","Design","operations"]] = {salary_series[["sales","design","operations"]]}')
```

```python
# Slicing
print(salary_series)
salary_series["hr":"support"]
```
