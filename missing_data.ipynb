```python
import numpy as np
import pandas as pd

departments = np.array(["HR", "Sales", "Tech", "Finance", "Marketing", "Support", "Design", "Legal", "Admin", "R&D"])
salaries = np.array([55000, np.nan, 72000, 68000, np.nan, 55000, 61000, 58000, np.nan, 72000])
salary_series = pd.Series (salaries, index=departments, name="Salary") 
salary_series
```

```python
salary_series + 2000
```

```python
# salary_series.isna() # Give True or False value Null values only
# salary_series.isna().sum() # Give total no. of Null values
salary_series[salary_series.isna()] # Give actual Null values
```

```python
salary_series.notna() # Give True or False Non_Null values only
salary_series[salary_series.notna()] # Give actual Non-Null values
```

```python
salary_series.isna().sum() # Total no.of Null values
```

```python
salary_series.hasnans # Give Null values availabe or not
```

```python
salary_series.value_counts() # Values and their apperances no.of times
```

```python
salary_series.value_counts(dropna = False) # Values and their apperance no.of times with Null also include
```

```python
salary_series.dropna() # Remove NUll values
salary_series # Till same (unchanged) orginal series
```

```python
salary_series.dropna().reset_index() # give dataframe values(sno,index,salary [3 columns]) and also remove Null Values
```

```python
salary_series.dropna().reset_index(drop = True) # remove index column
```

```python
salary_series.fillna(0) # Fill 0 everywhere exist Null value
```

```python
salary_series.fillna(salary_series.mean())
# salary_series.fillna(round(salary_series.mean(),2))
```
