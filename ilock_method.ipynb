```python
import pandas as pd
salaries = [45000, 52000, 0, 61000, 58000, 72000, 49000, 65000]
departments = ["developer", "marketing", "hr", "sales", "finance", "support", "design", "operations"]
salary_series = pd.Series (salaries, index = departments, name="Salary")


print(salary_series)
```

```python

# Indexing 
print("salary_series.iloc[0] = ",salary_series.iloc[0])
print("salary_series.iloc[2] = ",salary_series.iloc[2])
print("salary_series.iloc[-1] = ",salary_series.iloc[-1])
print("salary_series.iloc[1:4] = ",salary_series.iloc[1:4])
print("salary_series.iloc[[0,3,5]] = ",salary_series.iloc[[0,3,5]])
```
