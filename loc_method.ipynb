```python
import pandas as pd
salaries = [45000, 52000, 0, 61000, 58000, 72000, 49000, 65000]
departments = ["developer", "marketing", "hr", "sales", "finance", "support", "design", "operations"]
salary_series = pd.Series(salaries, index = departments , name = "Salary")
salary_series
```

```python
# Indexing
print('salary_series.loc["developer"] =',salary_series.loc["developer"])
print('salary_series.loc["hr"] =',salary_series.loc["hr"])
print('salary_series.loc["support"] =',salary_series.loc["support"])
print('salary_series.loc["marketing":"sales"] =',salary_series.loc["marketing":"sales"])
print('salary_series.loc[["developer","sales","support"]] =',salary_series.loc[["developer","sales","support"]])
```

```python
# slicing
print(salary_series)
print('salary_series.loc[: "hr"] = ', salary_series.loc[: "hr"])
print('salary_series.loc["sales":] = ', salary_series.loc["sales":])
print('salary_series.loc["developer":"finance"] = ', salary_series.loc["developer":"finance"])
print('salary_series.loc[salary_series > 50000] = ', salary_series.loc[salary_series > 50000])
print('salary_series.loc[salary_series == 0] = ', salary_series.loc[salary_series == 0])
 
```

```python
salary_series.loc['hr'] = 35000
salary_series
```

```python
print('salary_series.loc["marketing"] = ', salary_series.loc["marketing"])
print('salary_series.iloc[1] = ', salary_series.iloc[1])
```
