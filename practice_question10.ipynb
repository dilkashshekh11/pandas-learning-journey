```python

# Scenario: You're working with daily customer billing amounts. Your task is to apply discounts
# based on the billing amounts using different Pandas techniques.
import pandas as pd
billing = pd.Series ([450.75, 1200.50, 980.00, 1550.25,760.40, 2300.00, 890.10, 1475.60, 650.00, 3100.75],name="billing")
billing
```

```python

# 1.Write a function that gives 10% discount when billing is above 1000; otherwise return the same value.
print(billing)
final_price = billing.where(billing <= 1000,(billing * (0.90)))
final_price
```

```python

# 2.Use apply() to create a discounted Series from that function.
print(billing)
def final_amount(bill): #( I supposed 5% discount)
    if bill >= 2000:
        return bill - (bill *10/100)  # 10% discount
    elif bill >= 1000:
        return bill - (bill * 5/100) # 5% discount
    else:
          return bill               # No discount
    
billing.apply(final_amount)
```

```python

# 3.Use where() to create the same discounted Series without using apply().
final_result = billing.where(billing <= 1000, (billing * (0.90)))  # 10% discount
final_result
```

```python

# 4.Use the operator with where() to create the same result again.
final_result = final_result.where(final_result <= 1000, (final_result * (0.95))) # 10% discount
final_result
```

```python

# 5.Create a final Series where bills above 2000 get 15% dis bills above 1000 and up to 2000 get 10% discount, and the rest stay unchanged.
def final_bill(bill):
    if bill > 2000:
        return bill * 15/100
    elif bill > 2000:
        return bill * 10/100
    else:
        return bill
billing.apply(final_bill)
    
```
